# Notebook 07 — LSTM & Transformer ETA Models

Sequential ETA prediction for multi-hop delivery routes.
- LSTM captures downstream congestion propagation
- Temporal Transformer with time-of-day positional encoding
- Training curves and benchmark metrics

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import torch
import matplotlib.pyplot as plt
from src.models.lstm_eta import train as train_lstm
from src.models.transformer_eta import train as train_transformer
print(f'PyTorch {torch.__version__} | CUDA: {torch.cuda.is_available()}')
```

```python
df = pd.read_parquet('data/processed/features.parquet')
print(f'Dataset: {df.shape}')
print(df[['osrm_time','actual_time','dwell_time_proxy','corridor_mean_delay']].describe())
```

```python
# Train LSTM
lstm_result = train_lstm(df, epochs=40, batch_size=64)
print('LSTM result:', lstm_result)
```

```python
# Train Transformer
transformer_result = train_transformer(df, epochs=50, batch_size=64)
print('Transformer result:', transformer_result)
```

```python
# Save comparison
seq_results = pd.DataFrame([lstm_result, transformer_result])
seq_results.to_csv('data/processed/seq_model_results.csv', index=False)
print(seq_results)
```
